```python
salary_data.groupby("department")["salary"].agg(["count", "min","max","mean"]).round(2)
```

```python
# Store your result first
result = salary_data.groupby("department")["salary"].agg(["size" ,"min","max","mean"]).round(2)

# Then create a new gap column
result["gap"] = result["max"] - result["min"]

# Then sort by gap
result.sort_values("gap", ascending=False)
```
