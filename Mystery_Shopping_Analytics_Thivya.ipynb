##### Question 1 — Brand Performance Overview:
• Calculate average score per brand across ALL 10 parameters
• Create an overall_score = mean of all 10 parameter scores
• Rank brands from highest to lowest overall score
• Business question: Which brand delivers the best customer experience?
 

```python
score_cols = [
    'greeting_score',
    'product_knowledge_score',
    'store_cleanliness_score',
    'staff_appearance_score',
    'billing_efficiency_score',
    'customer_engagement_score',
    'complaint_handling_score',
    'upselling_score',
    'queue_management_score',
    'overall_experience_score'
]
```

```python
# Step 1: Create overall score for EACH AUDIT ROW

Mystery_shop_data["overall_score"] = (
    Mystery_shop_data[score_cols]
    .mean(axis=1).round(2)
)
Mystery_shop_data["overall_score"]
```

```python
# Step 2: Calculate average overall score for EACH BRAND
Overall_average = (
    Mystery_shop_data
    .groupby("brand")["overall_score"]
    .mean()
    .reset_index(name="Overall_average")
    .round(2)
)

Overall_average
```

```python
Overall_average = Overall_average.sort_values(
    by="Overall_average",
    ascending=False)

Overall_average
```

Which brand delivers the best customer experience? : Calvin Klein with an Overall CX average of 7.52

##### ## Section 7 — Key Findings & Recommendations
---

### 🏆 Key Findings

**Finding 1 — Brand Performance:**
Calvin Klein leads with highest overall CX score of 8.19
H&M and Zara are consistent underperformers across 
multiple parameters and cities.

**Finding 2 — Weakest Parameters Company-Wide:**
Three parameters consistently score below average:
- Complaint Handling: 6.82 — staff lack conflict resolution skills
- Product Knowledge: 6.82 — staff cannot answer customer queries
- Overall Experience: 6.84 — reflects cumulative service gaps

**Finding 3 — Geography Risk:**
Delhi, Hyderabad, and Kolkata require urgent intervention.
These cities show below-average scores across multiple brands
suggesting regional training gaps — not brand-specific issues.

**Finding 4 — Store-Level Risk:**
STR001 and STR002 are bottom performing stores.
Both require immediate store-level audit and action plan.

**Finding 5 — Announced vs Unannounced Gap:**
Scores are slightly higher during announced visits.
This indicates staff are not maintaining consistent 
service standards — they perform better only when 
they know they are being evaluated.

**Finding 6 — Manager Presence:**
Interestingly, scores are marginally higher when 
manager is absent. Staff appear to work more naturally 
and confidently without supervisory pressure.
Manager presence alone does not guarantee 
better service quality.

**Finding 7 — Performance Trend:**
Overall performance has marginally declined from 
6.87 (2023) to 6.86 (2024). While the decline is small,
the direction is concerning — intervention is needed
before it becomes a significant drop.

---

### 🎯 Recommendations

**Recommendation 1 — IMMEDIATE (Within 1 Month):**
Deploy targeted store audits at STR001 and STR002.
These bottom-performing stores need a physical 
inspection, staff interview, and a 30-day 
improvement plan with weekly check-ins.

**Recommendation 2 — SHORT TERM (Within 3 Months):**
Launch complaint handling and product knowledge 
training programs across all H&M and Zara stores 
in Delhi, Hyderabad, and Kolkata — the three cities 
with lowest performance scores.
These two skills account for the widest scoring gaps.

**Recommendation 3 — MEDIUM TERM (Within 6 Months):**
Increase frequency of unannounced mystery shopping 
visits from current ratio to 70% unannounced, 30% 
announced. This will give a more accurate picture 
of true daily service standards and reduce the 
performance gap between announced and unannounced visits.

**Recommendation 4 — BRAND SPECIFIC:**
H&M and Zara brand managers should review their 
staff training curriculum. Calvin Klein's training 
approach — which produces the highest CX scores — 
should be benchmarked and adapted for H&M and Zara.

**Recommendation 5 — MANAGEMENT APPROACH:**
Rather than increasing supervisory presence on 
shop floors, invest in building staff confidence 
and autonomy. Data shows staff perform slightly 
better without direct supervision — suggesting 
empowerment-based management works better than 
control-based supervision in retail environments.

**Recommendation 6 — TREND ALERT:**
The year-on-year decline — though marginal — 
must be arrested immediately. A quarterly 
performance review meeting between HS Brands 
and each brand manager is recommended to track 
progress against these recommendations.

---

### 💼 Overall Conclusion

> "Calvin Klein leads retail CX performance in India. 
> H&M and Zara require urgent brand-level intervention, 
> particularly in Delhi, Hyderabad, and Kolkata. 
> Staff training in complaint handling and product
