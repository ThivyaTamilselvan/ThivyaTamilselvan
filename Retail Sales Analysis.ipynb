#### Date & Time Analysis!

```python
Sales_data["Order Date"]
```

```python
# Convert text to datetime
Sales_data["Order Date"] = pd.to_datetime(
    Sales_data["Order Date"], dayfirst=True)
```

```python
# Extract date parts using .dt
Sales_data["Order Year"]    = Sales_data["Order Date"].dt.year
Sales_data["Order Month"]   = Sales_data["Order Date"].dt.month
Sales_data["Order Quarter"] = Sales_data["Order Date"].dt.quarter
```

```python
# Calculate date difference
Sales_data["Ship Date"] = pd.to_datetime(
    Sales_data["Ship Date"], dayfirst=True)
```

```python
Sales_data["Ship Date"]
```

```python
Sales_data["Shipping Days"] = (
    Sales_data["Ship Date"] - 
    Sales_data["Order Date"]
).dt.days
```

```python
Sales_data["Shipping Days"]
```

```python
#Total Sales by Year
Sales_data.groupby("Order Year")["Sales"].sum().round(2).reset_index().sort_values("Sales",ascending=False)
```

```python
#Total Sales by Month
Sales_data.groupby("Order Month")["Sales"].sum().round(2).reset_index().sort_values("Sales",ascending=False)
```

```python
#Total Sales by Quarter
Sales_data.groupby("Order Quarter")["Sales"].sum().round(2).reset_index().sort_values("Sales",ascending=False)
```

```python
## Section — Date & Time Analysis Findings

### Key Findings:

**Finding 1 — Year on Year Trend:**
2023 was peak year at $1.34M.
2024 shows concerning decline to $1.21M.
Root cause investigation recommended.

**Finding 2 — Quarterly Pattern:**
Q1 is strongest quarter at $1.03M — 
driven by Home Office segment new year purchases.
Q3 is weakest at $870K — summer slowdown.

**Finding 3 — Peak Month:**
January leads at $396K — Home Office segment 
drives new year purchases, not Corporate as expected.

**Finding 4 — Shipping Performance:**
All shipping modes deliver within 0.18 days of each other.
First Class premium pricing not justified by speed.
Recommend shipping pricing review.

**Finding 5 — Category Shipping:**
Office Supplies fastest at 3.88 days.
Furniture & Technology tied at 4.03 days
due to specialist handling requirements.
```

```python
# Prepare data for Chart 1 — Sales by Category
category_sales = Sales_data.groupby(
    "Category")["Sales"].sum().round(2).reset_index()

print(category_sales)
```

```python
# Chart 1 — Sales by Category
plt.figure(figsize=(5, 3))

plt.bar(
    category_sales["Category"],  # X axis — category names
    category_sales["Sales"],     # Y axis — sales values
    color=["steelblue", "orange", "green"]  # bar colours
)

plt.title("Total Sales by Category", fontsize=10)
plt.xlabel("Category", fontsize=10)
plt.ylabel("Total Sales ($)", fontsize=10)
# Formats Y axis with commas
plt.gca().yaxis.set_major_formatter(
    plt.FuncFormatter(lambda x, p: f"${x:,.0f}"))

plt.tight_layout()
plt.show()
```

```python
# Preparing data — Monthly Sales
monthly_sales = Sales_data.groupby(
    "Order Month")["Sales"].sum().round(2).reset_index()

print(monthly_sales)
```

```python
# Chart 2 — Monthly Sales Trend
plt.figure(figsize=(10, 3))

plt.plot(
    monthly_sales["Order Month"],  # X — months 1-12
    monthly_sales["Sales"],        # Y — sales values
    color="steelblue",             # line colour
    linewidth=2.5,                 # line thickness
    marker="o",                    # dot at each data point
    markersize=8                   # dot size
)

# Add value labels on each point
for x, y in zip(monthly_sales["Order Month"], 
                monthly_sales["Sales"]):
    plt.annotate(f"${y:,.0f}", 
                xy=(x, y),
                xytext=(0, 10),
                textcoords="offset points",
                ha="center",
                fontsize=8)

plt.title("Monthly Sales Trend", fontsize=10)
plt.xlabel("Month", fontsize=10)
plt.ylabel("Total Sales ($)", fontsize=10)

# Show month names instead of numbers
plt.xticks(
    range(1, 13),
    ["Jan","Feb","Mar","Apr","May","Jun",
     "Jul","Aug","Sep","Oct","Nov","Dec"]
)

plt.gca().yaxis.set_major_formatter(
    plt.FuncFormatter(lambda x, p: f"${x:,.0f}"))

plt.tight_layout()
plt.show()
```

```python
# Preparing data
region_sales = Sales_data.groupby(
    "Region")["Sales"].sum().round(2).reset_index().sort_values(
    "Sales", ascending=True)  # ascending for horizontal bar

print(region_sales)
```

```python
# Chart 3 — Sales by Region (Horizontal Bar)
plt.figure(figsize=(10, 3))

plt.barh(
    region_sales["Region"],   # Y axis — regions
    region_sales["Sales"],    # X axis — sales values
    color=["#2196F3","#4CAF50","#FF9800","#F44336"]
)

# Add value labels on each bar
for index, value in enumerate(region_sales["Sales"]):
    plt.text(
        value + 5000,          # X position
        index,                 # Y position
        f"${value:,.0f}",     # Label text
        va="center",           # Vertical alignment
        fontsize=10
    )

plt.title("Total Sales by Region", fontsize=10)
plt.xlabel("Total Sales ($)", fontsize=10)
plt.ylabel("Region", fontsize=10)

plt.gca().xaxis.set_major_formatter(
    plt.FuncFormatter(lambda x, p: f"${x:,.0f}"))

plt.tight_layout()
plt.show()
```

```python
# Preparing data
segment_profit = Sales_data.groupby(
    "Segment")["Profit"].sum().round(2).reset_index().sort_values(
    "Profit", ascending=False)

print(segment_profit)
```

```python
# Chart 4 — Profit by Segment
plt.figure(figsize=(10, 3))

# Colour bars based on profit value
colors = ["#2196F3","#4CAF50","#FF9800"]

bars = plt.bar(
    segment_profit["Segment"],
    segment_profit["Profit"],
    color=colors,
    width=0.5
)

# Add value labels ON TOP of each bar
for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width()/2,  # X — center of bar
        height + 500,                      # Y — just above bar
        f"${height:,.0f}",                # Label
        ha="center",                       # Horizontal align
        fontsize=11,
        fontweight="bold"
    )

plt.title("Total Profit by Customer Segment", fontsize=10)
plt.xlabel("Customer Segment", fontsize=10)
plt.ylabel("Total Profit ($)", fontsize=10)

plt.gca().yaxis.set_major_formatter(
    plt.FuncFormatter(lambda x, p: f"${x:,.0f}"))

plt.tight_layout()
plt.show()
```
